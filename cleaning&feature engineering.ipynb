```python
import pandas as pd 
import matplotlib.pyplot as plt 
import numpy as np 
```

Read DATA

```python
data = pd.read_csv(r'ml_repositories.csv')
data.head() 
```

```python
data.columns.tolist()
```

Start Cleaning Data

```python
data['id'] = data['id'].astype(str) 
data['created_at'] = pd.to_datetime(data['created_at']) 
data['updated_at'] = pd.to_datetime(data['updated_at']) 
data['language'] = data['language'].astype(str) 
```

```python
data['created_at'] = pd.to_datetime(data['created_at']).dt.tz_localize(None)
data['updated_at'] = pd.to_datetime(data['updated_at']).dt.tz_localize(None)

data['id'] = data['id'].astype(str)
data['language'] = data['language'].astype(str)
```

```python
data.head(10)
```

Sepreate time from created_at & uploaded_at columns

```python
data['created_at'] = pd.to_datetime(data['created_at'])
data['updated_at'] = pd.to_datetime(data['updated_at'])

data['time_updated_at'] = data['updated_at'].dt.time
data['time_created_at'] = data['created_at'].dt.time 

data['updated_at'] = data['updated_at'].dt.date
data['created_at'] = data['created_at'].dt.date 
```

```python
data.head(5)
```

```python
current_date = pd.to_datetime("2026-08-23", utc=True) 
data['repo_age_days'] = (current_date - pd.to_datetime(data['created_at'], utc=True)).dt.days 
```
